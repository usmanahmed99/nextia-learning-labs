## Setup: the helpers

The next cell defines three small functions that the notebook uses. You do not need to read them, but you can.

- `expect(what, yours, expected)` compares your number with the lesson's number. It says "Check passed." or if yours is too high or too low. It does not show the answer.
- `table(headers, rows)` prints a small table with aligned columns.
- `bars(labels, counts)` prints a text bar chart: one row of `█` for each label.

Run the cell. You should see `Ready.`

```python
# Helpers: check an answer without showing it, print a table, draw text bars.
import math, random, statistics

def expect(what, yours, expected, tol=0.01):
    """Compare your value with the expected one, to within `tol` for numbers."""
    if yours is ...:
        print(f"Not yet: replace ... with your calculation for {what}, then run the cell again.")
        return
    if isinstance(expected, (int, float)) and isinstance(yours, (int, float)):
        if abs(yours - expected) <= tol:
            print(f"Check passed: {what} is {yours:g}.")
        else:
            side = "too high" if yours > expected else "too low"
            print(f"Not yet: your {what}, {yours:g}, is {side}. Look at the step before it again.")
    elif yours == expected:
        print(f"Check passed: {what} is {yours}.")
    else:
        print(f"Not yet: your {what} is {yours!r}. That is not the expected answer.")

def table(headers, rows, digits=2):
    """Print rows as a table. Floats are rounded to `digits` decimals."""
    cell = lambda v: f"{v:,.{digits}f}" if isinstance(v, float) else str(v)
    text = [[cell(v) for v in row] for row in rows]
    widths = [max(len(str(h)), *(len(r[i]) for r in text)) for i, h in enumerate(headers)]
    print("  ".join(str(h).rjust(w) for h, w in zip(headers, widths)))
    for r in text:
        print("  ".join(v.rjust(w) for v, w in zip(r, widths)))

def bars(labels, counts, width=40):
    """Print a text bar chart: the longest bar is `width` characters."""
    top = max(counts) or 1
    size = max(len(str(label)) for label in labels)
    for label, n in zip(labels, counts):
        print(f"{str(label).rjust(size)} | {'█' * round(n / top * width)} {n}")

print("Ready.")
```

## 1. Two summaries of the centre, on three numbers

Run the cell. You should see `mean 20` and `median 20`.

```python
three = [10, 20, 30]
print("mean  ", statistics.mean(three))
print("median", statistics.median(three))
```

## 2. Monday's 12 tickets

These are the first-reply times, in minutes, of the 12 tickets T-2041 to T-2052 from one Monday morning. T-2052 arrived during a system outage. Run the cell. You should see `sum 345`, `mean 28.75` and `median 23.0`.

```python
tickets = ["T-2041", "T-2042", "T-2043", "T-2044", "T-2045", "T-2046",
           "T-2047", "T-2048", "T-2049", "T-2050", "T-2051", "T-2052"]
minutes = [12, 15, 18, 18, 20, 22, 24, 25, 27, 31, 38, 95]

print("count ", len(minutes))
print("sum   ", sum(minutes))
print("mean  ", statistics.mean(minutes))
print("median", statistics.median(minutes))
```

> **Check.** The median is halfway between the 6th and 7th values, (22 + 24) ÷ 2 = 23. The median is `23.0`, with a decimal point, because Python divides to get it.

The mean is not a typical wait here. Count the tickets that waited longer than the mean. Run the cell. You should see `3 of 12`.

```python
mean = statistics.mean(minutes)
above = [m for m in minutes if m > mean]
print(len(above), "of", len(minutes), "tickets waited longer than the mean:", above)
```

## 3. Move one ticket

On the lesson page, the explorer lets you move T-2052. Here you do the same with a list. The cell gives T-2052 a new wait and prints the mean, the median and the SD. (The **standard deviation (SD)** is a typical distance from the mean. Section 5 calculates it step by step.)

> **Predict.** T-2052 waits 195 minutes instead of 95. What happens to the mean? To the median? To the SD? Write it down, then run the cell.

```python
def summary(values):
    return round(statistics.mean(values), 2), statistics.median(values), round(statistics.pstdev(values), 2)

for t2052 in [95, 195]:
    mean, median, sd = summary(minutes[:11] + [t2052])
    print(f"T-2052 = {t2052:3} minutes: mean {mean}, median {median}, SD {sd}")
```

> **Check.** You should see a mean of 37.08, a median of 23.0 and an SD of 48.1 for 195 minutes. The extra 100 minutes add 100 ÷ 12 = 8.33 minutes to the mean. The median does not move, because T-2052 is still the largest value.

> **Your turn: change one thing.** In the cell above, change `[95, 195]` to `[95, 195, 0, 25]` and run it again. At 0 minutes, T-2052 is the smallest value: you should see a mean of 20.83, a median of 21.0 and an SD of 9.25. At 25 minutes, the mean and the median are both close to 23.

## 4. Percentiles: where a value stands

The lesson finds a percentile with a position rule:

1. Calculate the position: 1 + (n − 1) × p, where p is the percentile as a decimal.
2. If the position is a whole number, take that value.
3. If not, go between the two values around it, by the decimal part of the position.

The next cell writes the rule as a function and prints the lesson's table. Run it. You should see Q1 18, the median 23, Q3 28 and p90 37.3.

```python
def percentile(values, p):
    """The lesson's position rule: the same method as PERCENTILE.INC and numpy."""
    ordered = sorted(values)
    position = 1 + (len(ordered) - 1) * p
    below = math.floor(position)
    part = position - below
    if part == 0:
        return ordered[below - 1]
    return ordered[below - 1] + part * (ordered[below] - ordered[below - 1])

rows = []
for name, p in [("Q1 (p25)", 0.25), ("Median (p50)", 0.5), ("Q3 (p75)", 0.75), ("p90", 0.9)]:
    rows.append([name, round(1 + 11 * p, 2), round(percentile(minutes, p), 2)])
table(["Summary", "Position", "Minutes"], rows)
```

Python's `statistics.quantiles` gives the quartiles with the same method when you add `method="inclusive"`. Without it, Python uses another method, and small lists give slightly different values.

Run the cell. You should see `[18.0, 23.0, 28.0]`, then `[18.0, 23.0, 30.0]`: the default method gives another Q3.

```python
print(statistics.quantiles(minutes, n=4, method="inclusive"))
print(statistics.quantiles(minutes, n=4))   # the default method: different on 12 values
```

The **interquartile range (IQR)** is Q3 − Q1. Run the cell. You should see `IQR 10.0`.

```python
q1, median, q3 = statistics.quantiles(minutes, n=4, method="inclusive")
print("IQR", q3 - q1)
```

## 5. Spread: range, variance and standard deviation

The **range** is the largest value minus the smallest. The **variance** is the mean of the squared distances from the mean. The **SD** is its square root, back in minutes.

The next cell makes Priya's table: each distance from the mean, and its square.

> **Predict.** What is the sum of the distances? You saw it in the last lesson. Then run the cell.

```python
mean = statistics.mean(minutes)
rows = [[t, m, m - mean, (m - mean) ** 2] for t, m in zip(tickets, minutes)]
table(["Ticket", "Minutes", "Distance", "Squared"], rows)
print()
print("sum of distances:        ", sum(m - mean for m in minutes))
print("sum of squared distances:", sum((m - mean) ** 2 for m in minutes))
```

> **Check.** You should see a sum of distances of `0.0` and a sum of squared distances of `5342.25`. The distances cancel, so the variance squares them first.

Now divide by n, then take the square root. Run the cell. You should see a variance of 445.19, an SD of 21.1 and a range of 83.

```python
squares = sum((m - mean) ** 2 for m in minutes)
variance = squares / len(minutes)
print("variance (minutes²)", round(variance, 2))
print("SD (minutes)       ", round(math.sqrt(variance), 2))
print("same with pstdev   ", round(statistics.pstdev(minutes), 2))
print("range (minutes)    ", max(minutes) - min(minutes))
```

The lesson's column chart shows each ticket's share of the 5,342.25. Here is the same chart as text bars.

> **Predict.** What share of the total comes from T-2052? Then run the cell.

```python
shares = [round((m - mean) ** 2 / squares * 100, 1) for m in minutes]
bars([f"{t} ({m} min)" for t, m in zip(tickets, minutes)], shares)
```

> **Check.** You should see one long bar: T-2052 gives **82.2%** of the total. Squaring makes a large distance count much more than a small one.

**n or n − 1?** `statistics.pstdev` divides by n, as the lesson does. `statistics.stdev` (and pandas) divide by n − 1. Run the cell. You should see 21.1 and 22.04.

```python
print("divide by n:    ", round(statistics.pstdev(minutes), 2))
print("divide by n - 1:", round(statistics.stdev(minutes), 2))
```

## 8. Independent variation: Tuesday

On Tuesday morning, eight tickets have these first-reply times: 10, 12, 15, 15, 18, 20, 24 and 60 minutes.

> **Your turn.** Calculate the mean, the median, the range and the IQR. Then count the tickets that waited less than the mean. Run the check cell.

```python
tuesday = [10, 12, 15, 15, 18, 20, 24, 60]

mean_tue = statistics.mean(tuesday)
median_tue = statistics.median(tuesday)
range_tue = max(tuesday) - min(tuesday)
q1_tue, _, q3_tue = statistics.quantiles(tuesday, n=4, method="inclusive")
iqr_tue = q3_tue - q1_tue
below_mean_tue = len([m for m in tuesday if m < mean_tue])
```

## 9. Failure case: only the mean latency

Tomás reports that the ticket classifier answers in a mean time of 0.3 seconds. The support tool waits up to 3 seconds for a suggestion. Agents say that suggestions are often missing.

The next cell makes 500 answer times that match the lesson's description: most answers take 0.2 seconds, and 1 in 50 takes 5 seconds. These numbers are made up for this example.

> **Predict.** What are the mean, p50, p95 and p99? How many answers are over the 3-second limit? Then run the cell.

```python
seconds = [0.2] * 490 + [5.0] * 10   # 1 in 50 answers is slow

print("mean", round(statistics.mean(seconds), 3))
for name, p in [("p50", 0.5), ("p95", 0.95), ("p99", 0.99)]:
    print(name, round(percentile(seconds, p), 2))
over = len([s for s in seconds if s > 3])
print(f"over 3 seconds: {over} of {len(seconds)} ({over / len(seconds):.0%})")
```

> **Check.** You should see a mean of 0.296 (about 0.3), p50 and p95 of 0.2, and p99 of 5.0. Ten of 500 answers (2%) are over the limit. The mean and p95 look fine; only p99 and the count over the limit show the problem.

**Fix:** for any speed or waiting time, report p50, p95 and p99 with the mean, and the share over the limit that matters.
