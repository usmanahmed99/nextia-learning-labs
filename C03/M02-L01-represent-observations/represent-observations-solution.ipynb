## Setup: the helpers

The next cell defines three small functions that the notebook uses. You do not need to read them, but you can.

- `expect(what, yours, expected)` compares your number with the lesson's number. It says "Check passed." or if yours is too high or too low. It does not show the answer.
- `table(headers, rows)` prints a small table with aligned columns.
- `bars(labels, counts)` prints a text bar chart: one row of `█` for each label.

Run the cell. You should see `Ready.`

```python
# Helpers: check an answer without showing it, print a table, draw text bars.
import math, random, statistics

def expect(what, yours, expected, tol=0.01):
    """Compare your value with the expected one, to within `tol` for numbers."""
    if yours is ...:
        print(f"Not yet: replace ... with your calculation for {what}, then run the cell again.")
        return
    if isinstance(expected, (int, float)) and isinstance(yours, (int, float)):
        if abs(yours - expected) <= tol:
            print(f"Check passed: {what} is {yours:g}.")
        else:
            side = "too high" if yours > expected else "too low"
            print(f"Not yet: your {what}, {yours:g}, is {side}. Look at the step before it again.")
    elif yours == expected:
        print(f"Check passed: {what} is {yours}.")
    else:
        print(f"Not yet: your {what} is {yours!r}. That is not the expected answer.")

def table(headers, rows, digits=2):
    """Print rows as a table. Floats are rounded to `digits` decimals."""
    cell = lambda v: f"{v:,.{digits}f}" if isinstance(v, float) else str(v)
    text = [[cell(v) for v in row] for row in rows]
    widths = [max(len(str(h)), *(len(r[i]) for r in text)) for i, h in enumerate(headers)]
    print("  ".join(str(h).rjust(w) for h, w in zip(headers, widths)))
    for r in text:
        print("  ".join(v.rjust(w) for v, w in zip(r, widths)))

def bars(labels, counts, width=40):
    """Print a text bar chart: the longest bar is `width` characters."""
    top = max(counts) or 1
    size = max(len(str(label)) for label in labels)
    for label, n in zip(labels, counts):
        print(f"{str(label).rjust(size)} | {'█' * round(n / top * width)} {n}")

print("Ready.")
```

Run the cell. You should see `T-4101: [2, 2, 0, 1]` and `dimension 4`.

## 2. From many tickets to a table of numbers

Priya writes three tickets one under the other: one row for each ticket, one column for each feature. In plain Python, a **matrix** is a list of rows. Its **shape** is (rows, columns).

Run the cell. You should see the three rows and `shape (3, 4)`.

```python
tickets = ["T-4101", "T-4102", "T-4103"]
X = [
    [2, 2, 0, 1],   # T-4101
    [1, 0, 3, 0],   # T-4102
    [3, 1, 1, 1],   # T-4103
]

def shape(matrix):
    """(number of rows, number of columns) of a list of rows."""
    return (len(matrix), len(matrix[0]))

for row in X:
    print(row)
print("shape", shape(X))
```

## 4. Worked example: Priya builds the feature matrix

Priya starts from the ticket texts. A short function counts the refund words: "refund", "charged" and "money back". The texts are the starts of the texts from the lesson's table.

Run the cell. You should see the counts `2`, `0` and `1`.

```python
REFUND_WORDS = ["refund", "charged", "money back"]

def refund_words(text):
    """How often "refund", "charged" or "money back" appears in the text."""
    text = text.lower()
    return sum(text.count(word) for word in REFUND_WORDS)

texts = {
    "T-4101": "I was charged for a lamp that arrived broken. Please refund me…",
    "T-4102": "Where is my order? This is the third time I write…",
    "T-4103": "The garden chair has a crack. Can I get my money back or a new one…",
}
for ticket, text in texts.items():
    print(ticket, refund_words(text))
```

Now Priya does the lesson's four steps: convert each value to the feature's unit, write each ticket as a vector in the fixed order, stack the vectors as rows, and check the shape.

Run the cell. You should see `T-4101 [2.0, 2, 0, 1]` and two more rows, then `shape (3, 4)` and `same as X? True`.

```python
raw = [
    # ticket,  words, past tickets, photo
    ("T-4101", 200, 0, "yes"),
    ("T-4102", 100, 3, "no"),
    ("T-4103", 300, 1, "yes"),
]

rows = []
for ticket, words, past, photo in raw:
    vector = [words / 100, refund_words(texts[ticket]), past, 1 if photo == "yes" else 0]
    rows.append(vector)
    print(ticket, vector)

print("shape", shape(rows))
print("same as X?", rows == X)
```

> **Check.** Python prints the length as `2.0`, `1.0` and `3.0`, because `/` always gives a decimal number. `2.0 == 2` is true in Python, so the matrix is the same as X.

A professional keeps this matrix in a numpy array or a pandas DataFrame. Then `X.shape` shows `(3, 4)`.

## 7. Failure case: the columns change order

After a pipeline change, the columns arrive in a different order: past tickets first, then length. The shape stays the same, so a shape check does not catch it. The model still reads position 1 as length.

The fix in the lesson: compare simple summaries with last week. The next cell calculates the mean of each column of the five tickets, first in the correct order, then in the changed order.

> **Predict.** Which column means change places? Then run the cell.

```python
X5 = [[2, 2, 0, 1], [1, 0, 3, 0], [3, 1, 1, 1], [4, 2, 2, 0], [1, 0, 0, 0]]
moved = [[row[2], row[1], row[0], row[3]] for row in X5]   # past tickets first, then length

def column_means(matrix):
    return [statistics.mean(column) for column in zip(*matrix)]

print("shapes:", shape(X5), shape(moved))
table(["Position", "Expected feature", "Mean before", "Mean after"],
      [[i + 1, f, float(b), float(a)] for i, (f, b, a) in enumerate(zip(features, column_means(X5), column_means(moved)))])
```

> **Check.** You should see two equal shapes, `(5, 4)`. The mean in position 1 falls from 2.20 to 1.20, and the mean in position 3 rises from 1.20 to 2.20. "Length" now has the mean of "past tickets". No error message appears.

**Fix:** keep the feature names with the numbers, and select the columns by name before the model scores. Run the cell. You should see the five rows in the correct order again and `correct again? True`.

```python
# The pipeline sends each ticket with its column names.
named = [dict(zip(["past tickets", "refund words", "length", "photo"], row)) for row in moved]
fixed = [[ticket[name] for name in features] for ticket in named]
for row in fixed:
    print(row)
print("correct again?", fixed == X5)
```
