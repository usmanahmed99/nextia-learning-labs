## Setup: the helpers

The next cell defines small functions that this notebook uses. You do not need to read them, but you can.

- `check("""assert ...""")` runs the `assert` lines of a check cell. It prints `Check passed.`, or `Not yet:` and what is different. It does not show the answer.

Run the cell. It prints nothing.

```python
# Helpers: run the project, and check an answer without showing it.
def check(asserts):
    """Run the assert lines of a check. Say what is different, never the answer."""
    try:
        exec(asserts, globals())
    except AssertionError as error:
        print(f"Not yet: {error}")
    except NameError as error:
        print(f"Not yet: {error}. Run your cell above first.")
    except Exception as error:
        print(f"Not yet: the check stopped with {type(error).__name__}: {error}")
    else:
        print("Check passed.")
```

## 1. One ticket, five kinds of value

Here is one support ticket, as a person writes it down:

> Ticket T-1042, priority 2, open for 3.5 hours, urgent: yes, assigned to: nobody yet.

In Python, each fact is a **value**, and each value has a **type**. The type decides what you can do with the value. A **variable** is a name for a value. The `=` sign means "let this name refer to this value".

Run the next cell. Each line shows a value and its type.

```python
ticket_id = "T-1042"
priority = 2
hours_open = 3.5
is_urgent = True
assigned_to = None

print(ticket_id, type(ticket_id))
print(priority, type(priority))
print(hours_open, type(hours_open))
print(is_urgent, type(is_urgent))
print(assigned_to, type(assigned_to))
```

> **Check.** You should see five lines, with the types `str` (text), `int` (a whole number), `float` (a number with a decimal point), `bool` (`True` or `False`) and `NoneType` ("no value").

> **Your turn: change one thing.** In the cell above, change `priority = 2` to `priority = "2"`, and run it again. What is the type of `priority` now? The quotes make it text, although it looks like a number.

## 2. A variable can change

When you use `=` again, the name refers to the new value. Python calculates the right side first, then gives the name the result.

> **Predict.** Write down the value of `hours_open` after each line. Then run the cell.

```python
hours_open = 3.5
hours_open = hours_open + 2
print("after line 2:", hours_open)
hours_open = hours_open * 2
print("after line 4:", hours_open)
```

> **Check.** You should see `5.5` and then `11.0`. A table with the value after each line is a **state table**. Make one when code does something that you do not expect.

## 3. Numbers

`/` divides. `//` gives the whole-number part of a division, and `%` gives the remainder. `round(value, 2)` rounds to 2 decimal places.

> **Predict.** What do `7 / 2`, `7 // 2`, `7 % 2` and `6 / 2` give? Then run the cell.

```python
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(6 / 2)
print(round(3.14159, 2))
print(0.1 + 0.2)
```

> **Check.** You should see `3.5`, `3`, `1`, `3.0`, `3.14` and `0.30000000000000004`. Division with `/` always gives a `float`. A `float` is stored in binary, so some decimal numbers are not exact: use `round()` when you show the result.

## 4. Text

A **string** is text in quotes. Strings have **methods**: actions that you call with a dot after the value. `strip()` removes the spaces at the start and the end, and `lower()` makes all the letters small. A method gives you a new string. It does not change the original.

An **f-string** has `f` before the quote. Python replaces each name in `{ }` with its value. Run the cell.

```python
raw_status = "  Open "
status = raw_status.strip().lower()   # the text after # is a comment
print(repr(raw_status), "->", repr(status))

ticket_id = "T-1042"
priority = 2
print(f"Ticket {ticket_id} has priority {priority}.")
```

> **Check.** You should see `'  Open ' -> 'open'` and `Ticket T-1042 has priority 2.` (`repr()` shows a string with its quotes, so that you can see the spaces.)

## 5. True, False and None

A comparison gives `True` or `False`. `==` means "is equal to". (One `=` gives a name to a value.) `and`, `or` and `not` combine conditions. To check for `None`, use `is None`.

> **Predict.** Which lines print `True`? Then run the cell.

```python
priority = 2
hours_open = 3.5
is_urgent = True
assigned_to = None

print(priority == 2)
print(priority != 2)
print(hours_open > 24)
print(priority == 1 or hours_open > 24)
print(is_urgent and assigned_to is None)
print(not is_urgent)
```

> **Check.** You should see `True`, `False`, `False`, `False`, `True`, `False`.

`None` is different from `0` and from the empty string `""`. A ticket with priority `0` has a priority. A ticket with priority `None` does not.

## 6. Convert text to numbers

When Python reads a file, every value arrives as a string, even the numbers. You cannot calculate with `"3.5"`. Here is what happens when you try.

> **Predict.** What does `24 - "3.5"` do? Then run the cell.

The `try` and `except` lines catch the error, so that the notebook continues. You learn them in Module 5.

```python
hours_text = "3.5"
try:
    print(24 - hours_text)
except TypeError as error:
    print("TypeError:", error)
```

> **Check.** You should see `TypeError: unsupported operand type(s) for -: 'int' and 'str'`. Python does not guess that the text is a number.

Convert it first: `int("2")` gives `2`, `float("3.5")` gives `3.5`, and `str(2)` gives `"2"`. `int()` accepts only whole-number text: `int("2.5")` and `int("three")` are errors.

> **Your turn.** The two values in the next cell came from a file, so they are text. Change the two conversion lines, so that `priority` is an `int` and `hours_open` is a `float`. Then run the check cell. It calculates the hours until the 24-hour deadline and shows `Priority 2: 20.5 hours left`.

```python
# These values came from a file, so they are text.
priority_text = "2"
hours_text = "3.5"

# Convert them.
priority = int(priority_text)
hours_open = float(hours_text)
```

```python
check("""
assert type(priority) is int, "priority must be an int. Use int()."
assert type(hours_open) is float, "hours_open must be a float. Use float()."
hours_left = 24 - hours_open
assert hours_left == 20.5, f"hours_left is {hours_left!r}. It must be 20.5."
print(f"Priority {priority}: {hours_left} hours left")
""")
```

## 7. Your turn: hours to days

Python accepts almost any name, but a good name says what the value is and its unit: `hours_open`, not `h` or `x`. A name that starts with `is_` reads as a yes/no question. Python's style is small letters with `_` between words.

> **Your turn.** A ticket has been open for `"50"` hours. The value is text, because it came from a file. Write code that shows this line:
>
> `T-1042 has been open for 2 days and 2 hours.`
>
> Use `//` and `%`. Give each variable a name that says what it is and its unit. The check looks for an `int` variable whose name ends with `days`.

```python
ticket_id = "T-1042"
hours_text = "50"

total_hours = int(hours_text)
full_days = total_hours // 24
extra_hours = total_hours % 24

print(f"{ticket_id} has been open for {full_days} days and {extra_hours} hours.")
```

```python
check("""
assert any(v == 2 for k, v in dict(globals()).items() if k.endswith("days") and type(v) is int), "Make a variable whose name ends with days, with the value 2."
""")
```

## 8. Failure case: text plus a number

Tomás wants to show the priority in a sentence, and joins the text and the number with `+`.

> **Predict.** What happens? Then run the cell.

```python
priority = 2
try:
    print("Priority: " + priority)
except TypeError as error:
    print("TypeError:", error)
```

> **Check.** You should see `TypeError: can only concatenate str (not "int") to str`. `+` adds two numbers, or joins two strings. It cannot join a string to a number. **Concatenate** means "join text together".

**Fix:** decide which type you need. To show a number in text, use an f-string. To calculate with text that holds a number, convert it first. If you are not sure of a type, print `type(value)`. Run the cell.

```python
priority = 2
priority_text = "2"
print(f"Priority: {priority}")
print(int(priority_text) + 2)
print(type(priority_text))
```

> **Check.** You should see `Priority: 2`, `4` and `<class 'str'>`.
