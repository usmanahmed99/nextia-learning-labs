## Setup: the helpers

The next cell defines small functions that this notebook uses. You do not need to read them, but you can.

- `check("""assert ...""")` runs the `assert` lines of a check cell. It prints `Check passed.`, or `Not yet:` and what is different. It does not show the answer.

Run the cell. It prints nothing.

```python
# Helpers: run the project, and check an answer without showing it.
def check(asserts):
    """Run the assert lines of a check. Say what is different, never the answer."""
    try:
        exec(asserts, globals())
    except AssertionError as error:
        print(f"Not yet: {error}")
    except NameError as error:
        print(f"Not yet: {error}. Run your cell above first.")
    except Exception as error:
        print(f"Not yet: the check stopped with {type(error).__name__}: {error}")
    else:
        print("Check passed.")
```

## 1. Lists: items in order

A **collection** is one value that holds other values. A **list** keeps items in order, in square brackets. Each item has a position, its **index**, which starts at `0`: the index counts how many items come before the item.

> **Predict.** What do `statuses[0]`, `statuses[-1]` and `statuses[1:3]` give? Then run the cell.

```python
statuses = ["open", "closed", "open", "pending"]

print(statuses[0])
print(statuses[1])
print(statuses[-1])
print(len(statuses))
print(statuses[1:3])
statuses.append("open")
print(len(statuses))
```

> **Check.** You should see `open`, `closed`, `pending`, `4`, `['closed', 'open']` and `5`. `-1` is the last item. A **slice** `[1:3]` goes from index 1 up to, but not including, index 3. `append` adds an item at the end.

A `for` loop runs the indented lines once for each item. The `for` line ends with `:`, and the lines inside the loop are indented by four spaces. Run the cell.

```python
for status in statuses:
    print(status)
```

## 2. Dictionaries: values by name

A **dictionary** finds a value by a **key**, a name that you choose. Each entry is a key and a value, with `:` between them. A dictionary fits a record: `ticket["priority"]` explains itself.

> **Predict.** What does `ticket.get("assignee")` give, for a key that does not exist? Then run the cell.

```python
ticket = {"id": "T-1001", "status": "open", "category": "billing", "priority": 2}

print(ticket["category"])
print(ticket.get("assignee"))
print(ticket.get("assignee", "nobody"))
ticket["status"] = "closed"

for key, value in ticket.items():
    print(key, "=", value)
```

> **Check.** You should see `billing`, `None`, `nobody`, then the four keys and values, with `status = closed`. `ticket["assignee"]` with square brackets stops with `KeyError: 'assignee'`. `.get()` gives `None`, or your default, instead.

## 3. Tuples and sets

A **tuple**, in round brackets, is a small group of values that cannot change. You can **unpack** it into separate names. A **set** keeps each value only once, and has no order. `in` checks if a value is there. Run the cell.

```python
pair = ("T-1001", "open")
ticket_id, status = pair
print(ticket_id, status)

categories = set(["billing", "login", "billing", "shipping"])
print(sorted(categories))
print("login" in categories)
```

> **Check.** You should see `T-1001 open`, `['billing', 'login', 'shipping']` and `True`. A set has no order, so use `sorted()` when you need a fixed order.

## 4. Worked example: a list of ticket records

Most of the work in this course uses a **list of dictionaries**: each dictionary is one ticket, and the list holds all the tickets. Here are the six tickets that the rest of this module uses.

> **Predict.** Write down the first three lines of the output before you run the cell.

```python
tickets = [
    {"id": "T-1001", "status": "open", "category": "billing", "priority": 2},
    {"id": "T-1002", "status": "closed", "category": "login", "priority": 3},
    {"id": "T-1003", "status": "open", "category": "login", "priority": 1},
    {"id": "T-1004", "status": "pending", "category": "billing", "priority": 2},
    {"id": "T-1005", "status": "open", "category": "billing", "priority": 3},
    {"id": "T-1006", "status": "closed", "category": "shipping", "priority": 1},
]

print("Number of tickets:", len(tickets))

first = tickets[0]
print("First ticket:", first["id"], first["status"])

print("Category of the last ticket:", tickets[-1]["category"])

for ticket in tickets:
    print(ticket["id"], "-", ticket["status"])
```

## 5. Guided practice: answer questions about the tickets

Steps 1 and 2 are done for you. `set()` makes an empty set, and `.add()` puts one value in it.

> **Your turn.** Do steps 3 and 4 in the cell:
>
> 3. Set `has_refund` to `True` or `False`: does any ticket have the category `"refund"`? Use `in`.
> 4. Change the status of the first ticket to `"closed"`, then show the whole first ticket.

> Then run the check cell.

```python
tickets = [
    {"id": "T-1001", "status": "open", "category": "billing", "priority": 2},
    {"id": "T-1002", "status": "closed", "category": "login", "priority": 3},
    {"id": "T-1003", "status": "open", "category": "login", "priority": 1},
    {"id": "T-1004", "status": "pending", "category": "billing", "priority": 2},
    {"id": "T-1005", "status": "open", "category": "billing", "priority": 3},
    {"id": "T-1006", "status": "closed", "category": "shipping", "priority": 1},
]

# 1. Status of the third ticket
print(tickets[2]["status"])

# 2. All categories, sorted
categories = set()
for ticket in tickets:
    categories.add(ticket["category"])
print(sorted(categories))

# 3. Is there a "refund" ticket?
has_refund = "refund" in categories
print(has_refund)

# 4. Close the first ticket, then show it.
tickets[0]["status"] = "closed"
print(tickets[0])
```

## 6. Two names, one list

`=` does not copy a value. It gives the value another name.

> **Predict.** Write down the three numbers that you expect. Then run the cell.

```python
tickets = ["T-1001", "T-1002"]
backup = tickets
copied = tickets.copy()

tickets.append("T-1003")

print("tickets:", len(tickets))
print("backup: ", len(backup))
print("copied: ", len(copied))
```

> **Check.** You should see `3`, `3` and `2`. There is only one list for `tickets` and `backup`: `backup = tickets` made a second name for it, so a change through one name is visible through the other. `.copy()` makes a new list.

> **Warning.** `.copy()` copies the list, but not the dictionaries in it. Run the cell: a change to `backup[0]["status"]` also changes `tickets[0]["status"]`. In Module 3, you write functions that make new records instead of changing old ones.

```python
tickets = [{"id": "T-1001", "status": "open"}]
backup = tickets.copy()
backup[0]["status"] = "closed"
print(tickets[0]["status"])
```

> **Check.** You should see `closed`.

## 7. Your turn: a lookup by ID

A colleague asks for tickets by ID, many times a day.

> **Your turn.** Make a dictionary `by_id` whose keys are the ticket IDs and whose values are the ticket records. Start with an empty dictionary, and add one entry for each ticket with a loop. Then `by_id["T-1004"]["status"]` gives `"pending"`. Run the check cell after it.

```python
tickets = [
    {"id": "T-1001", "status": "open", "category": "billing", "priority": 2},
    {"id": "T-1002", "status": "closed", "category": "login", "priority": 3},
    {"id": "T-1003", "status": "open", "category": "login", "priority": 1},
    {"id": "T-1004", "status": "pending", "category": "billing", "priority": 2},
    {"id": "T-1005", "status": "open", "category": "billing", "priority": 3},
    {"id": "T-1006", "status": "closed", "category": "shipping", "priority": 1},
]

by_id = {}
for ticket in tickets:
    by_id[ticket["id"]] = ticket

print(by_id["T-1004"]["status"])
```

```python
check("""
assert len(by_id) == 6, f"by_id has {len(by_id)} entries. It must have 6, one for each ticket."
assert by_id["T-1004"]["status"] == "pending", "by_id['T-1004'] must be the record of ticket T-1004."
""")
```

A lookup by key is fast, even with a million tickets, because a dictionary does not search through the items one by one.

## 8. Failure case: IndexError and KeyError

You ask for an item that is not there. A list with 6 items has the indexes `0` to `5`. The `try` and `except` lines catch each error, so that the notebook continues (Module 5 explains them).

> **Predict.** Which error does each line give? Then run the cell.

```python
ticket = tickets[0]
try:
    print(tickets[6])
except IndexError as error:
    print("IndexError:", error)
try:
    print(ticket["Status"])
except KeyError as error:
    print("KeyError:", error)
```

> **Check.** You should see `IndexError: list index out of range` and `KeyError: 'Status'`. The key in the message shows what Python looked for: `"Status"` and `"status"` are different keys.

**Fix:** print what is there: `len(tickets)` for a list, `ticket.keys()` for a dictionary. Use an index below `len(...)`. For a key that can be missing, use `.get()`. Run the cell.

```python
print(len(tickets))
print(ticket.keys())
print(ticket.get("assignee"))
```

> **Check.** You should see `6`, `dict_keys(['id', 'status', 'category', 'priority'])` and `None`.
