## 1. Anatomy of a function

A **function** is a named block of code that you can run many times. `def` starts the definition: Python remembers the function, but does not run the body yet. The name in brackets is the **parameter**. The **docstring** says what the function gives back. `return` ends the function and gives a value back to the caller.

Run the next cell. It makes the six tickets of this module, and defines `count_by_category`. It prints nothing.

```python
tickets = [
    {"id": "T-1001", "status": "open", "category": "billing", "priority": 2},
    {"id": "T-1002", "status": "closed", "category": "login", "priority": 3},
    {"id": "T-1003", "status": "open", "category": "login", "priority": 1},
    {"id": "T-1004", "status": "pending", "category": "billing", "priority": 2},
    {"id": "T-1005", "status": "open", "category": "billing", "priority": 3},
    {"id": "T-1006", "status": "closed", "category": "shipping", "priority": 1},
]


def count_by_category(tickets):
    """Return the number of tickets in each category."""
    counts = {}
    for ticket in tickets:
        category = ticket["category"]
        counts[category] = counts.get(category, 0) + 1
    return counts
```

To **call** the function, write its name with a value in brackets, the **argument**. The call is replaced by the returned value. The same function runs three times with different arguments. Run the cell.

```python
all_counts = count_by_category(tickets)
print("All tickets:", all_counts)
print("Billing tickets:", all_counts["billing"])
print("No tickets:", count_by_category([]))
```

> **Check.** You should see `All tickets: {'billing': 3, 'login': 2, 'shipping': 1}`, `Billing tickets: 3` and `No tickets: {}`.

The variables in the body, such as `counts`, are **local**: they exist only while the function runs. The only way to get a value out is `return`.

> **Predict.** What happens when you print `counts` outside the function? Then run the cell. (`try` and `except` catch the error, so that the notebook continues. Module 5 explains them.)

```python
try:
    print(counts)
except NameError as error:
    print("NameError:", error)
```

## 2. The contract

A function's **contract** is the promise that it makes to the code that calls it. Write it in plain words before you write the body:

| Question | `count_by_category` |
|---|---|
| What goes in? | A list of ticket dictionaries. Each one has a `"category"` key. |
| What comes out? | A dictionary: category → number of tickets. Empty input gives `{}`. |
| What does it **not** do? | It does not change the list, and it does not print. |

A function that only reads its input and returns a result is a **pure function**. Pure functions are the easiest to check: give an input, compare the output.

## 4. Defaults

A parameter can have a **default value**. The caller can leave out that argument. A named argument, `status="closed"`, makes a call easier to read. Run the cell.

```python
def filter_by_status(tickets, status="open"):
    """Return the tickets that have the given status."""
    selected = []
    for ticket in tickets:
        if ticket["status"] == status:
            selected.append(ticket)
    return selected


print(len(filter_by_status(tickets)))
print(len(filter_by_status(tickets, "closed")))
print(len(filter_by_status(tickets, status="closed")))
```

> **Check.** You should see `3`, `2` and `2`.

## 5. Worked example: count_tickets.py as functions

Here is the Module 2 script, rewritten as functions. Each function has one job. `main()` calls them in order, and it is the only function that prints. `TICKETS` is in capital letters: a **constant**. `average_priority` returns `None` for empty input, instead of stopping with `ZeroDivisionError`. `return` with no value ends `main()` early.

Run the cell. The last line calls `main()`.

```python
TICKETS = [
    {"id": "T-1001", "status": "open", "category": "billing", "priority": 2},
    {"id": "T-1002", "status": "closed", "category": "login", "priority": 3},
    {"id": "T-1003", "status": "open", "category": "login", "priority": 1},
    {"id": "T-1004", "status": "pending", "category": "billing", "priority": 2},
    {"id": "T-1005", "status": "open", "category": "billing", "priority": 3},
    {"id": "T-1006", "status": "closed", "category": "shipping", "priority": 1},
]


def filter_by_status(tickets, status="open"):
    """Return the tickets that have the given status."""
    selected = []
    for ticket in tickets:
        if ticket["status"] == status:
            selected.append(ticket)
    return selected


def count_by_category(tickets):
    """Return the number of tickets in each category, sorted by category."""
    counts = {}
    for ticket in tickets:
        category = ticket["category"]
        counts[category] = counts.get(category, 0) + 1
    return dict(sorted(counts.items()))


def average_priority(tickets):
    """Return the average priority, or None when there are no tickets."""
    if not tickets:
        return None
    total = 0
    for ticket in tickets:
        total = total + ticket["priority"]
    return round(total / len(tickets), 1)


def main():
    status = "open"
    selected = filter_by_status(TICKETS, status)
    print(f"Tickets with status '{status}': {len(selected)}")
    if not selected:
        print("No tickets to count.")
        return
    for category, count in count_by_category(selected).items():
        print(f"  {category}: {count}")
    print(f"Average priority: {average_priority(selected)}")


main()
```

> **Check.** You should see `Tickets with status 'open': 3`, `  billing: 2`, `  login: 1` and `Average priority: 2.0`: the same output as the script of Module 2. That is the test of a **refactor**: the structure changed, and the behaviour did not.

## 6. Check each function alone

Each function is pure, so a few lines can check it. `assert` stops with an error if a condition is `False`, and does nothing if it is `True`. Run the cell.

```python
two = [
    {"id": "A", "status": "open", "priority": 1},
    {"id": "B", "status": "closed", "priority": 2},
]

assert filter_by_status(two) == [two[0]]
assert filter_by_status(two, "closed") == [two[1]]
assert filter_by_status([]) == []
assert average_priority(two) == 1.5
print("All checks passed.")
```

## 8. Failure case: a default list that remembers

A function gives a different result each time that you call it with the same arguments.

> **Predict.** The second call should return only `['unknown status']`. What does it return? Then run the cell.

```python
def add_problem(problem, problems=[]):
    problems.append(problem)
    return problems

print(add_problem("missing id"))
print(add_problem("unknown status"))
```

> **Check.** You should see `['missing id']`, then `['missing id', 'unknown status']`. Python makes the default value **once**, when it runs the `def` line, not on each call. A list can change, so every call without the argument gets the same list. This is the "two names, one list" problem from [Collections](https://learning.nextia-ai.com/courses/python/m02/collections/), in a less visible place.

**Fix:** use `None` as the default, and make a new list in the body. Run the cell. You should see `['missing id']`, then `['unknown status']`.

```python
def add_problem(problem, problems=None):
    if problems is None:
        problems = []
    problems.append(problem)
    return problems

print(add_problem("missing id"))
print(add_problem("unknown status"))
```

Use only values that cannot change as defaults: numbers, strings, `True`, `False` and `None`. Never a list, a dictionary or a set.
